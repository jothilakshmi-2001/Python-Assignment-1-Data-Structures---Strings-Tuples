# 2. String Slicing and Indexing:

```python
text = "Hello Jothi, welcome to Python programming"
print(text)
```

```python
#  a. Print the first character of the string
print("a. First character:", text[0])
```

```python
# b. Print the last character of the string
print("b. Last character:", text[-1])
```

```python
# c. Print the first 5 characters of the string
print("c. First 5 characters:", text[:5])
```

```python
# d. Print the last 11 characters of the string
print("d. Last 11 characters:", text[-11:])
```

```python
# e. Print the string in reverse
print("e. Reversed string:", text[::-1])
```

```python
# f. Use slicing and print the word “Python” from the existing string
print("f. Extracted word:", text[24:30])
```

# 3. String Methods:

```python
strM = "Python beginner tutorial"
print(strM)
```

```python
# a. Convert the sentence to uppercase
print("a. Uppercase:", strM.upper())
```

```python
# b. Convert the sentence to lowercase
print("b. Lowercase:", strM.lower())
```

```python
# c. Use Capitalize (makes the first letter uppercase)
print("c. Capitalize:", strM.capitalize())
```

```python
# d. Count the total number of occurrences of character ‘t’
print("d. Count of 't':", strM.count('t'))
```

```python
# e. Replace "Python" with "Machine Learning"
print("e. Replaced string:", strM.replace("Python", "Machine Learning"))
```

# Tuples (Creation, Modification and Access) :

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
```

```python
# a. Concatenate the two tuples and store it in “t_combine”
t_combine = tuple1 + tuple2
print("a. t_combine:", t_combine)
```

```python
# b. Repeat the elements of “t_combine” 3 times
t_repeat = t_combine * 3
print("b. t_repeat:", t_repeat)
```

```python
# c. Access the 3rd element from “t_combine” (Index starts at 0, so 3rd element is at index 2)
print("c. 3rd element:", t_combine[2])
```

```python
# d. Access the first three elements from “t_combine”
print("d. First three elements:", t_combine[0:3])
```

```python
# e. Access the last three elements from “t_combine”
print("e. Last three elements:", t_combine[-3:])
```
